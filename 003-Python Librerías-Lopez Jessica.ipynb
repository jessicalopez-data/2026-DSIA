Ejercicio 2: Dataset simple con Pandas

estudiante	nota

```
import pandas as pd
datos=pd.DataFrame({"estudiante":["Ana","Juan","Carla","Pedro","Lucia"],"nota":[8,6,9,5,7]})
print(datos)
```

```
print("Promedio:",datos["nota"].mean())
print("Nota maxima:",datos["nota"].max())
print("Nota minima:",datos["nota"].min())
```
